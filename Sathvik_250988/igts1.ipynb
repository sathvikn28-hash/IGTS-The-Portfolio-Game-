```python
def diff(lst,x):
    seen=[]
    for num in lst:
        if (num+x) in seen:
            return(num,num+x)
        if (num-x) in seen:
            return(num,num-x)
        seen.append(num)
lst=[67,45,32,100,33,42,50]
x=67
print("Pair with difference ",x," is ",diff(lst,x))
```

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(0)
daily_returns = np.random.normal(loc=0.0008, scale=0.015, size=252)
price_series = 100 * np.cumprod(1 + daily_returns)
plt.figure(figsize=(12, 5))
plt.plot(price_series, color='steelblue', linewidth=1.5)
plt.title('Simulated Price Series (252 Trading Days)')
plt.xlabel('Trading Day')
plt.ylabel('Price')
plt.show()
final_price   = price_series[-1]
peak_price    = price_series.max()
rolling_peak  = np.maximum.accumulate(price_series)
drawdowns     = (price_series - rolling_peak) / rolling_peak * 100
max_drawdown  = drawdowns.min()
print("Final Price: ",final_price)
print("Peak Price: ",peak_price)
print("Maximum Drawdown: ",max_drawdown)
```

```python
import matplotlib.pyplot as plt
months = ['Jan','Feb','Mar','Apr','May','Jun',
          'Jul','Aug','Sep','Oct','Nov','Dec']
year1  = np.array([42, 45, 50, 48, 55, 60, 58, 63, 67, 70, 75, 80])
year2  = np.array([52, 49, 58, 61, 66, 71, 68, 74, 79, 83, 88, 95])
plt.figure(figsize=(12,8))
plt.subplot(2,1,1)
plt.plot(months,year1,label='Year1',color='royalblue')
plt.plot(months,year2,label='Year2',color='darkorange')
plt.title('Monthly Revenue(Year1 vs Year2)')
plt.xlabel('Months')
plt.ylabel('Revenue (in Lakhs)')
plt.legend()
diff=year2-year1
colors=[]
for i in diff:
    if i<0:
        colors.append('red')
    else: 
        colors.append('green')
plt.subplot(2,1,2)
plt.bar(months,diff,color=colors)
plt.show()
```

```python
import numpy as np, pandas as pd
df = pd.DataFrame({
'Day': range(1, 11),
'Price': [100, 103, np.nan, 108, 107, np.nan, 112, 115, np.nan, 120],
'Volume': [500, 520, 510, np.nan, 530, 545, np.nan, 560, 570, 580]
})
empty_price = df['Price'].isnull().sum()
empty_volume = df['Volume'].isnull().sum()
print('Empty values in Price: ',empty_price)
print('Empty values in Volume: ',empty_volume)
df['Price'] = df['Price'].ffill()
df['Volume'] = df['Volume'].fillna(df['Volume'].mean())
df['Return'] = (df['Price'] - df['Price'].shift(1)) / df['Price'].shift(1) * 100
print("DataFrame with Return column:")
print(df)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
np.random.seed(42)
returns = np.random.multivariate_normal(
    mean=[0.0005, 0.0008, 0.0003],
    cov=[[0.0004, 0.0001, 0.00005],
         [0.0001, 0.0006, 0.00008],
         [0.00005, 0.00008, 0.0003]],
    size=200
)
prices = 100 * np.cumprod(1 + returns, axis=0)

df_prices = pd.DataFrame(prices, columns=['Asset_A', 'Asset_B', 'Asset_C'])
df_prices['Portfolio'] = (1/3)*(df_prices['Asset_A']+df_prices['Asset_B']+df_prices['Asset_C'])

plt.figure(figsize=(12, 6))
plt.plot(df_prices['Asset_A'],   label='Asset A',  color='blue')
plt.plot(df_prices['Asset_B'],   label='Asset B',  color='orange')
plt.plot(df_prices['Asset_C'],   label='Asset C',  color='green')
plt.plot(df_prices['Portfolio'], label='Equal-Weighted Portfolio',
         color='black', linewidth=2)
plt.title('Multi-Asset Price Series & Equal-Weighted Portfolio')
plt.xlabel('Trading Day')
plt.ylabel('Price')
plt.legend()
plt.show()

df_returns=pd.DataFrame(returns,columns=['Asset_A', 'Asset_B', 'Asset_C'])
print("Correlation Matrix:")
print(df_returns.corr())
```
